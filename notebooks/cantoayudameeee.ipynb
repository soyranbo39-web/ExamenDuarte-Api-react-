```python

import requests
from pprint import pprint

API_URL = "http://127.0.0.1:8000"
```

```python

usuario = {
    "full_name": "Test Header",
    "username": "headeruser",
    "password": "headerpass123"
}

r = requests.post(f"{API_URL}/auth/usuarios", json=usuario)
pprint(r.json())
```

```python

login_data = {
    "username": "headeruser",
    "password": "headerpass123"
}

r = requests.post(f"{API_URL}/auth/login", json=login_data)
login_response = r.json()
pprint(login_response)

token = login_response["body"]["access_token"]
```

```python

headers = {"Authorization": f"Bearer {token}"}
r = requests.get(f"{API_URL}/auth/user/me", headers=headers)
pprint(r.json())
```

```python

user_data = r.json()
assert user_data["username"] == "headeruser"
assert user_data["name"] == "Test Header"
print("Autenticación por header exitosa")
```
